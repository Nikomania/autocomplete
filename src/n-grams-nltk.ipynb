# Utilização de N-GRAMAS com NLTK


# 

```python
import nltk

from nltk import word_tokenize
from nltk.util import ngrams
from nltk.lm import MLE
from nltk.lm.preprocessing import padded_everygram_pipeline
```

```python
frases = [
    "eu quero beber água",
    "eu quero beber café",
    "eu quero comer",
    "eu quero comer alguma coisa",
    "eu quero ir para casa",
    "eu quero ir ao banheiro",

    "eu estou com fome",
    "eu estou com sede",
    "eu estou cansado",
    "eu estou com dor",

    "preciso de ajuda",
    "preciso ir ao banheiro",
    "preciso beber água",

    "quero falar com minha mãe",
    "quero falar com meu pai",
]

# Modelo trigrama
N = 3
```

```python
# Tokenização
tokens = [
    word_tokenize(frase.lower(), language="portuguese")
    for frase in frases
]

train_data, vocabulario = padded_everygram_pipeline(
    N,
    tokens
)
```

## Treino de um modelo com as frases acima para trigramas

```python
model = MLE(N)

model.fit(train_data, vocabulario)
```

## Teste do modelo para autocompletar

```python
def next_word(model, frase, num_top_pred=5):
    tokens = word_tokenize(frase.lower(), language="portuguese")

    contexto = tokens[-2:]

    palavras = []

    for palavra in model.vocab:
        probabilidade = model.score(
            palavra,
            contexto
        )

        if probabilidade > 0:
            palavras.append(
                (palavra, probabilidade)
            )

    palavras.sort(
        key=lambda x: x[1],
        reverse=True
    )

    return palavras[:num_top_pred]
```

```python
print(next_word(model, "eu quero"))
```
